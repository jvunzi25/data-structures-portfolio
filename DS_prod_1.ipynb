```python
df = pd.DataFrame({'Team': ['Warriors','Bulls'],
                   'Points':[115.9,round(8625/82, 2)],
                   'Assist': [30.4, round(2033/82, 2)],
                   'Rebounds':[44.43, round(3658/82, 2)],
                   'Steals':[9.57, round(745/82, 2)],
                   'Blocks':[6.77, round(345/82, 2)],
                   'FG%':[.495,.478],
                   'Opp_FG%':[.435,.448]
                   })
df
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
stats = df.melt(
    id_vars="Team",
    var_name="Statistic",
    value_name="Value"
)

plt.figure(figsize=(12, 6))
sns.barplot(
    data=stats,
    x="Statistic",
    y="Value",
    hue="Team"
)

plt.title("Team Statistics Comparison")
plt.xlabel("Statistic")
plt.ylabel("Value")
plt.legend(title="Team")
plt.tight_layout()
plt.show()
```

```python
fg_stats = stats[stats["Statistic"].isin(["FG%", "Opp_FG%"])].copy()
fg_stats["Value"] = fg_stats["Value"] * 100

plt.figure(figsize=(8, 6))
sns.barplot(
    data=fg_stats,
    x="Statistic",
    y="Value",
    hue="Team"
)

plt.title("Field Goal Percentage Comparison")
plt.xlabel("Statistic")
plt.ylabel("Percentage (%)")
plt.ylim(0, 60)
plt.legend(title="Team")
plt.tight_layout()
plt.show()
```
